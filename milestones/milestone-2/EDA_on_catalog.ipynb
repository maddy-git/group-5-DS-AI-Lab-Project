#### Maping the images with SKU In dataset with image path

```python
image_root = r"C:\Users\z048724\notebook\Data Scince Lab\week - 4\product_dataset\cloth_data_for_embedding\images"
```

```python

image_map = {}
for root, _, files in os.walk(image_root):
    for file in files:
        if file.lower().endswith(('.jpg', '.jpeg', '.png')):
            fname = os.path.splitext(file)[0]
            normalized_sku = fname.lstrip('0')
            rel_path = os.path.relpath(os.path.join(root, file), start=os.path.dirname(image_root))
            rel_path = rel_path.replace("\\", "/")
            image_map[normalized_sku] = rel_path   # <- you also need to populate the dict!
            
cloth_data["Image_URL"] = (
    cloth_data["SKU_No"]
    .astype(str)
    .str.lstrip("0")
    .map(image_map)
)

cloth_data["Image_URL"] = cloth_data["Image_URL"].fillna("Not_Found")

print(cloth_data[["SKU_No", "Image_URL"]].head())
```

#### Sample Images

```python
sample_rows = cloth_data[cloth_data["Image_URL"] != "Not_Found"].head(5)



plt.figure(figsize=(15, 6))
for i, (sku, rel_path) in enumerate(zip(sample_rows["SKU_No"], sample_rows["Image_URL"]), start=1):
    image_path = os.path.join(os.path.dirname(image_root), rel_path)
    if os.path.exists(image_path):
        img = Image.open(image_path)
        plt.subplot(1, len(sample_rows), i)
        plt.imshow(img)
        plt.axis("off")
        plt.title(str(sku))
    else:
        print(f"File not found: {image_path}")

plt.tight_layout()
plt.show()
```
